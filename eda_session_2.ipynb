```python
continents = visa_df['continent'].unique()
count = []
for i in continents:
    con = visa_df['continent'] == i  # one class
    count.append( np.sum(con) )
count

data = zip(continents,count)
df = pd.DataFrame(data)
df.columns = ['continent','freq']
df
```

```python
continents = visa_df['continent'].unique()
freq=[]
for i in continents:
    cond = visa_df['continent'] == i
    freq.append(np.sum(cond))
freq
```
