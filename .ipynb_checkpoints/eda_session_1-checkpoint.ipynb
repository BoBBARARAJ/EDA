## important

### **`Dimensions`**

```python
[]  == 1d   series
[[]]  == 2d    df
[[[]]]  == 3d  
```

```python
# 1D and 2D

# List [] Numpy : 1D  Pandas  === series

# List [[]]  Numpy   : 2D  Pandas  === Dataframe
```

```python
# column name not there === series
# column name there ===   dataframe
```

```python
# column  = 1D  === series
# column  = 2D  === dataframe

visa_df

visa_df['continent']   # 1D  Series
visa_df[['continent']]   # 2D  df

visa_df.iloc[[100,200,300],3]  # 1D series
visa_df[[100,200,300],[3]]  # 2D df
visa_df.loc[[100,200,300],'continent']  # 1D series
visa_df[[100,200,300],['continent']]  # 2D df
```

```python
rows = [100,200,300]
visa_df.iloc[rows, 3]  # 1D series
visa_df.iloc[rows, [3]]  # 2D df
visa_df.loc[rows, ['continent']]  # 2D df
visa_df.loc[rows, 'continent']  # 1D series
```

## task-2


### seperate categorical columns and numerical columns

```python
keys = visa_df.dtypes.keys()
values = visa_df.dtypes.values
```

```python
categorical = []
numerical = []
for key,value in zip(keys,values):
    if value == 'object':
        categorical.append(key)
    else:
        numerical.append(key)
```

```python
print(categorical)
print()
print(numerical)
```

```python
data = [categorical,numerical]
df  = pd.DataFrame(data).T
df.columns = ['categorical' , 'numerical']
df
```
