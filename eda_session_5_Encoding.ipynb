## pd.get_dummies

```python
-  get dummies is a way to represent the categorical columns   to numerical 

-  first it will check what are the labels in a column 

- for ex : case status has  two labels

    - Denied and Certified  ddnb

- So it will create 2 new columns based on lables

    - col1  : case_staus_Denied

    - col2  : case_status_Certified


case_status     |    case_status_Denied   |   case_status_Certified   | ddnb

denied          |          1                            0

certified       |          0                             1

certified       |          0                             1

denied          |          1                              0

.
```
